```python
import json
```

```python
with open("train_bronze.json", "r") as file: 
    bronze = json.load(file)
with open("train_gold.json", "r") as file: 
    gold = json.load(file)
with open("train_silver.json", "r") as file: 
    silver = json.load(file)
with open("train_silver_2025.json", "r") as file: 
    silver2 = json.load(file)

data = {**gold, **silver, **silver2}
```

```python
from tqdm import tqdm
equivocado = 0
errores = {}
for pmid, entry in tqdm(data.items()):
    for entity1 in entry["entities"]:
        span1 = entity1["text_span"]
        label1 = entity1["label"]
        for pmid2, entry2 in bronze.items():
            for entity2 in entry2["entities"]:
                span2 = entity2["text_span"]
                label2 = entity2["label"]
                if span1 == span2 and label1 != label2:
                    equivocado += 1
                    if span1 not in errores: 
                        errores[span1] = (label1, label2)
                    else:
                        continue
                    #print(pmid, pmid2, span1, "  |  ", label1, "  |  ",  label2)
print(equivocado)

```

```python
cont = 0
print(len(errores))
for termino, error in errores.items():
    if cont<100:
        print(termino, error)
        cont += 1
        
```

```python
with open("errores_Bronze.json", "w") as file:
    json.dump(errores, file)
```
